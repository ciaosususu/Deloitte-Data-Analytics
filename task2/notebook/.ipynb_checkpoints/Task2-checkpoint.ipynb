```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_excel("../data/Equality Table.xlsx")
```

```python
def classify_equality(score):
    if -10 <= score <= 10:
        return "Fair"
    elif (-20 <= score < -10) or (10 < score <= 20):
        return "Unfair"
    else:
        return "Highly Discriminative"
```

```python
df["Equality Class"] = df["Equality Score"].apply(classify_equality)
```

```python
df.head()
```

```python
df.to_excel("../data/Equality Table - Classified.xlsx", index=False)
```

```python
print(df.columns.tolist())
```

```python
plt.figure(figsize=(8,6))
df["Equality Class"].value_counts().plot(kind='bar')
plt.title("Distribution of Equality Classes")
plt.ylabel("Number of Job Roles")
plt.xlabel("Equality Class")
plt.tight_layout()
plt.savefig("../images/equality_score_distribution.png", dpi=300)
plt.show()
```
